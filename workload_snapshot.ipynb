# What an 80% signal means in Swiss job adverts

Reproduce the 4 November 2026 aggregate from [Panta](https://withpanta.com/insights/eighty-percent-careers-switzerland/). This is observed advert wording, not national prevalence or an agreed four-day week.


```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd() / 'examples'))
from _public import get
data = get('/insights/data/eighty-percent-careers-2026-11-04.json')
c = data['main']['categories']
compatible = c['exact_80'] + c['range_including_80'] + c['discrete_80_option']
assert compatible == data['main']['includes_80']
print(data['snapshot_at'], data['main']['records'], compatible)
print(f"Ranges among compatible signals: {100*c['range_including_80']/compatible:.1f}%")
```

```python
from html import escape
rows = [('Exact 80%', c['exact_80']), ('Continuous range including 80%', c['range_including_80'])]
svg = ['<svg xmlns="http://www.w3.org/2000/svg" width="900" height="240" viewBox="0 0 900 240">', '<rect width="900" height="240" fill="white"/>', '<text x="20" y="30" font-size="20">80%-compatible advert signals, 4 November 2026</text>']
for i, (label, count) in enumerate(rows):
    y = 60 + i*60
    svg += [f'<text x="20" y="{y+20}">{escape(label)}</text>', f'<rect x="300" y="{y}" width="{count/compatible*480:.1f}" height="30" fill="#216d65"/>', f'<text x="800" y="{y+20}">{count:,}</text>']
svg += ['<text x="20" y="210">Panta: 3,365 detected compatible signals within 14,536 fresh-backed physical CH listings.</text>', '<text x="20" y="230">Title/employment fields; not national estimates or confirmed schedules. withpanta.com/insights/</text>', '</svg>']
Path('workload-signals.svg').write_text(''.join(svg))
```

## Limits

This notebook reproduces published aggregate arithmetic; it does not re-run the original extraction or independently validate semantic labels. Missing workload fields remain unknown. The 170 exact values plus 3,195 ranges equal 3,365 compatible signals; no unconflicted discrete alternative was detected. Study codebook, sensitivity and review limitations are on the original report. Run from the repository root.
